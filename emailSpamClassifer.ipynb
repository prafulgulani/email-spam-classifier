## Evaluation Methodology

To evaluate and compare the performance of statistical and embedding-based models in spam email classification, several key metrics are used. These metrics help assess not only how accurate the models are but also how well they handle imbalanced datasets, like the spam classification task, where one class (non-spam) is typically much larger than the other (spam). The following metrics are crucial for this comparison:

1. **Accuracy**  
   - **Definition**: Accuracy is the proportion of correctly classified instances (both spam and non-spam) out of the total instances.  
   - **Application**: This metric provides a general idea of how well each model is performing in overall classification. However, for imbalanced datasets like spam detection, accuracy alone can be misleading, as a model might achieve high accuracy by simply predicting the majority class (non-spam). It is important to look beyond accuracy to understand model performance.

     $\text{Accuracy} = \frac{\text{True Positives} + \text{True Negatives}}{\text{Total Samples}}$

2. **Precision**  
    - **Definition**: Precision measures the proportion of correctly identified spam emails among all emails predicted as spam.  
    - **Application**: Precision is crucial when false positives (non-spam classified as spam) must be minimized. For example, marking an important email as spam could result in a poor user experience. Embedding-based models like BERT are often more adept at understanding nuanced language and context, which can help improve precision compared to statistical models like Naive Bayes.

      $\text{Precision} = \frac{\text{True Positives (TP)}}{\text{True Positives (TP)} + \text{False Positives (FP)}}$  

3. **Recall**  
    - **Definition**: Recall measures the proportion of actual spam emails that were correctly identified.  
    - **Application**: Recall is essential when it is critical to minimize false negatives (spam classified as non-spam). Missing spam emails can leave users vulnerable to phishing or malicious content. Embedding-based models often achieve higher recall due to their ability to generalize better across diverse spam patterns, whereas statistical models might struggle with variations in spam language. 

      $\text{Recall} = \frac{\text{True Positives (TP)}}{\text{True Positives (TP)} + \text{False Negatives (FN)}}$  

4. **F1-Score**  
   - **Definition**: The F1-Score is the harmonic mean of precision and recall, providing a single metric that balances both.  
   - **Application**: This metric is especially useful when dealing with imbalanced datasets. It helps evaluate the trade-off between precision and recall. Embedding-based models tend to achieve a better F1-Score due to their ability to better capture the underlying context and relationships within the data, while statistical models may struggle in this regard.

     $\text{F1-Score} = 2 \times \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}$

By applying these metrics, I will share my insights about how well statistical model and embedding-based model perform for the task of spam email classification. This comparison will help in identifying which model is more suited for real-world applications where false positives or negatives might have significant consequences.

### Tokenization and Lemmatization for TF-IDF Representation

**What is Tokenization?**

Tokenization is the process of breaking down a text into smaller units, typically words or sentences. These units, called tokens, serve as the building blocks for text analysis.

**Why is Tokenization Important?**
- Text Preprocessing: It simplifies text into manageable chunks for processing.
- Input for NLP Models: Tokenized text is the foundation for downstream tasks like lemmatization, TF-IDF, and embeddings.
- Preserves Structure: Tokenization ensures that the sequence of words is maintained, which is crucial for tasks like sentiment analysis or language modeling.

**What is Lemmatization?**

Lemmatization reduces words to their base or root form, called a lemma. Unlike stemming, which may simply truncate words, lemmatization uses a vocabulary and grammar rules to produce valid words.

**Why is Lemmatization Important?**
- Reduces Redundancy: It treats different forms of a word (e.g., "run," "running," "ran") as a single entity.
- Improves Model Efficiency: By reducing dimensionality, lemmatization makes the data more interpretable and compact.
- Critical for TF-IDF and Vectorization: Lemmatization ensures that variations of the same word are not treated as separate features.

```python
# stop words, lemmatizer and punkt tokenizer
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
tokenizer = PunktLanguageVars()

def tokenize_and_lemmatize(text):
    # tokenize the text into words
    tokens = tokenizer.word_tokenize(text)

    # remove the stopwords and lemmatize each word
    lemmatized_tokens = [lemmatizer.lemmatize(token) for token in tokens if token not in stop_words]

    # join the cleaned words
    return ' '.join(lemmatized_tokens)
```

```python
# Applying the tokenize_and_lemmatize function to the clean_text
df['lemmatized_text'] = df['clean_text'].apply(tokenize_and_lemmatize)

# the first few rows of the final text
df[['clean_text', 'lemmatized_text']].head()
```

### TF-IDF Representation

#### What is TF-IDF?
- **TF-IDF (Term Frequency-Inverse Document Frequency)** is a statistical measure that evaluates the importance of a word in a document relative to a collection of documents. 
- It helps highlight significant terms while reducing the impact of commonly used words like "the" and "and".

#### Why Use TF-IDF?
- Converts text data into numerical form for machine learning algorithms.
- Captures the importance of words based on how frequently they appear in a document vs. how rare they are in the document.
- Often used for tasks like text classification, clustering, and information retrieval.

#### How Does It Work?
1. **Term Frequency (TF)**: Measures how often a word appears in a document.
   
   $TF = \frac{\text{Number of times a term appears in a document}}{\text{Total number of terms in the document}}$

2. **Inverse Document Frequency (IDF)**: Reduces the weight of commonly occurring words.
   
   $IDF = \log\left(\frac{\text{Total number of documents}}{\text{Number of documents containing the term}}\right)$
   
3. **TF-IDF Score**: The final score is computed as:
   
   $TF\text{-}IDF = TF \times IDF$

Word counts alone fail to capture the importance or rarity of terms in a document, which limits their utility in distinguishing informative words from common ones. TF-IDF addresses this by assigning higher weights to significant terms while reducing the influence of frequent, less meaningful ones, improving the performance of models like logistic regression, SVM, or Naive Bayes. However, it is less effective for deep learning models such as BERT, which rely on raw text and leverage contextual embeddings to capture word meanings.

```python
# TF-IDF vectorizer
tfidf_vectorizer = TfidfVectorizer(max_features=10000)

# Fitting the cleaned text
tfidf_matrix = tfidf_vectorizer.fit_transform(df['lemmatized_text'])

# Converting the TF-IDF matrix to a DataFrame
tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns=tfidf_vectorizer.get_feature_names_out())

tfidf_df.head()
```

### **BERT Tokenization**

BERT (Bidirectional Encoder Representations from Transformers) is a transformer-based model pre-trained on large corpora, making it adept at understanding language context. Unlike methods like TF-IDF or bag-of-words, BERT captures the semantic meaning of words and their contextual relationships within sentences.

#### **Why Use BERT Tokenization?**
BERT uses its own tokenizer because:
1. **Subword Tokenization**: Rare or unseen words are split into subwords ("untidy" -> "un", "tidy") to retain meaning and handle out-of-vocabulary words.
2. **Consistency**: Ensures compatibility with BERT's pre-trained embeddings.
3. **Preservation of Context**: Works on the raw text without manual lemmatization or stemming, preserving full semantic meaning.

#### **Key Features of BERT's Tokenization**
- **WordPiece Tokenizer**: BERT uses WordPiece tokenization, which splits words into subwords based on their frequency in the training corpus.
- **Special Tokens**:
  - `[CLS]`: Added at the beginning of the input as a placeholder for the entire sequence’s representation.
  - `[SEP]`: Indicates the end of a sentence or separates multiple sentences.
- **Maximum Sequence Length**: Typically 512 tokens; longer texts are truncated.

```python
# BERT tokenizer
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# Example text
texts = df['text'].tolist()

# Tokenize, pad, and create attention masks
input_ids = []
attention_masks = []

for text in texts:
    encoded_dict = tokenizer.encode_plus(
        text,
        add_special_tokens=True,     # add [CLS] and [SEP]
        max_length=128,              # sequence length
        truncation=True,
        padding='max_length',
        return_attention_mask=True,  # generate attention mask
        return_tensors='pt'          # return PyTorch tensors
    )
    
    input_ids.append(encoded_dict['input_ids'])
    attention_masks.append(encoded_dict['attention_mask'])

# Convert to tensors
input_ids = torch.cat(input_ids, dim=0)
attention_masks = torch.cat(attention_masks, dim=0)

print(f"Input IDs: {input_ids[:2]}")
print(f"Attention Masks: {attention_masks[:2]}")
```

#### Output
1. **Input IDs**: Numerical representations of tokens.
2. **Attention Masks**: Binary masks indicating which tokens should be attended to (1 for actual tokens, 0 for padding).


### Conclusion: Preprocessing for Statistical and Embedding Models

#### Statistical Model (Logistic regression using TF-IDF):

For the TF-IDF model, the primary goal was to convert the raw text into a numerical format suitable for machine learning algorithms like Logistic Regression. The steps taken included:

- **Text Cleaning**: I removed non-alphabetic characters and converted the text to lowercase to standardize the data.
- **Tokenization and Lemmatization**: Tokenization split the text into individual words, and lemmatization ensured that words were reduced to their base forms, minimizing variations and improving consistency.
- **TF-IDF Representation**: The processed text was then transformed using TF-IDF, which assigns a weight to each word based on its importance in the document relative to the entire dataset. This method captures the relevance of each word in the context of the email, providing a sparse but effective representation for the machine learning model.

#### Embedding Model (BERT):

For the BERT model, which uses pre-trained word embeddings, the approach is more sophisticated and involves:

- **Tokenization with BERT's Tokenizer**: Unlike TF-IDF, BERT uses its own tokenizer that splits the text into subword.
- **Padding and Attention Masks**: Given that BERT requires fixed-length input sequences, padding was applied to ensure that all texts had the same length. Attention masks were also created to inform the model which parts of the input are real tokens and which are padding.
- **Word Embeddings**: BERT uses contextual embeddings, meaning that the representation of each word depends on the surrounding words in the sentence, allowing it to capture semantic meaning more effectively than statistical methods like TF-IDF.

Overall, while the TF-IDF method focused on extracting important word-level features for the statistical model, BERT leveraged contextualized word embeddings to understand the deeper semantic meaning of the text. The preprocessing steps reflect the underlying differences in how these models handle text data: TF-IDF relies on frequency-based features, whereas BERT operates on more complex word embeddings that capture context and meaning.

In this dataset there is an imbalance, where 53% of the emails are spam and the remaining 47% are non-spam (ham) emails. While this imbalance isn't severe, it can still impact the performance of machine learning models. A model trained on imbalanced data may become biased towards the majority class (spam), leading to poor performance in identifying the minority class (ham).

To address this, several techniques can be applied:
1. **Resampling Techniques:**
   - Oversampling the minority class or undersampling the majority class can help balance the dataset. This can lead to more accurate predictions, especially for the underrepresented class.
   
2. **Class Weights:**
   - Some machine learning algorithms allow adjusting the weight of each class during training. By assigning higher weights to the minority class, the model can be made to pay more attention to correctly classifying the spam emails.

3. **Synthetic Data Generation:**
   - Techniques like SMOTE (Synthetic Minority Over-sampling Technique) can generate synthetic samples of the minority class to create a more balanced dataset without losing valuable data.

Given the relatively mild imbalance, resampling or class weighting should be sufficient to handle this issue without significantly altering the data distribution. So I will be using class weighting for both the methods as it can be easily applied while using the algorithms.

```python
# selection of device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
bert_model.to(device)

epochs = 3 # number of training epochs

for epoch in range(epochs):
    bert_model.train()
    total_loss = 0
    
    for step, batch in enumerate(train_dataloader):
        # move data to device
        batch_input_ids = batch[0].to(device)
        batch_input_mask = batch[1].to(device)
        batch_labels = batch[2].to(device)

        bert_model.zero_grad() # reset gradient
        outputs = bert_model(batch_input_ids, attention_mask=batch_input_mask, labels=batch_labels)
        loss = outputs.loss
        total_loss += loss.item() # total loss
        loss.backward()
        optimizer.step() # update parameters using optimizer

    # avg loss for epoch
    avg_loss = total_loss / len(train_dataloader)
    print(f"Epoch {epoch+1}: Loss = {avg_loss}")
```

```python
# evaluation
bert_model.eval()
predictions, true_labels = [], []

for batch in val_dataloader:
    batch_input_ids = batch[0].to(device)
    batch_input_mask = batch[1].to(device)
    batch_labels = batch[2].to(device)

    # no gradient to speed up evaluation
    with torch.no_grad():
        outputs = bert_model(batch_input_ids, attention_mask=batch_input_mask)
        logits = outputs.logits

    preds = torch.argmax(logits, dim=1).cpu().numpy()
    labels = batch_labels.cpu().numpy()
    
    predictions.extend(preds)
    true_labels.extend(labels)

accuracy = accuracy_score(true_labels, predictions)
precision = precision_score(true_labels, predictions)
recall = recall_score(true_labels, predictions)
f1 = f1_score(true_labels, predictions)

print(f"Accuracy: {accuracy}")
print(f"Precision: {precision}")
print(f"Recall: {recall}")
print(f"F1-Score: {f1}")
```
